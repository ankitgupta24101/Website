```python
import pandas as pd
import numpy as np
```

```python
# Step 1: Create sample dataset with 20 rows including missing values, duplicates, and outliers
```

```python
data = {
    'name': ['Aarav', 'Bhavya', 'Chirag', 'Deepa', 'Ekta', 'Farhan', 'Gaurav', 'Harshita', 'Ishaan', 'Jaya',
             'Kiran', 'Lalit', 'Manisha', 'Naveen', 'Ojas', 'Pooja', 'Anand', 'Ritika', 'Bhavya', 'Tanya'],
    'age': [25, 30, 35, np.nan, 45, 40, 35, 28, 30, 32,
            25, 30, 120, 40, 29, np.nan, 33, 310, 30, 28],  # includes missing values and an outlier (120)
    'salary': [50000, 60000, 55000, 70000, np.nan, 62000, 58000, 59000, np.nan, 61000,
               50000, 60000, 1000000, 62000, 59000, 60000, 610000, 59000, 60000, 58000],  # includes missing and outlier (1,000,000)
    'department': ['HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR',
                   'HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR', 'IT', 'IT', 'HR']
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original DataFrame:")
df
```

```python
# Step 2: Handle missing values
# Fill missing 'age' with median, 'salary' with median
```

```python
df['age'] = df['age'].fillna(df['age'].median())
df['salary'] = df['salary'].fillna(df['salary'].median())
```

```python
print("After Handling missing values:")
df
```

```python
# Step 3: Remove duplicates
```

```python
duplicates = df[df.duplicated(keep=False)]
print("Duplicate rows in the dataset:")
duplicates
```

```python
df = df.drop_duplicates()
```

```python
print("After Removing duplicates:")
df
```

```python
# Step 4: Type casting
```

```python
# Use .loc to explicitly assign values to the DataFrame that is a view or slice rather than a copy
```

```python
# df['age'] = df['age'].astype(int)
# df['salary'] = df['salary'].astype(int)
df.loc[:, 'age'] = df['age'].astype(int)
df.loc[:, 'salary'] = df['salary'].astype(int)
```

```python
print("After Type casting:")
df
```

```python
# Step 5: Handle outliers using IQR for 'age' and 'salary'
```

```python
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
```

```python
# Apply IQR Method
df = remove_outliers(df, 'age')
df = remove_outliers(df, 'salary')
```

```python
print(remove_outliers(df, 'age'))
print(remove_outliers(df, 'salary'))
```

```python
print("After Handling outliers:")
df.reset_index(drop=True)
```
