```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sna
```

```python
data = {
    "Students": ["A","B","C","D","E","F"],
    "Math":[78,85,67,96,56,89],
    "Science":[82,79,45,90,87,82],
    "English" :[45,78,98,25,78,67]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# linePlot
# https://matplotlib.org/stable/api/markers_api.html
```

```python
plt.plot(df["Students"],df["Math"],marker = "o", label ="Math")
plt.plot(df["Students"],df["Science"],marker = "s", label ="Science")
plt.plot(df["Students"],df["English"],marker = "v", label ="English")
plt.title("Line Plot")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.show()
```

```python
 # Bar Chart
```

```python
plt.bar(df["Students"], df["Math"], color = "skyblue")
plt.title("Bar Chart")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
# Histogram
```

```python
plt.hist(df["Science"], bins = 5, color = "purple", edgecolor = "black")
plt.title("Histogram")
plt.xlabel("Marks Range")
plt.ylabel("Frequency")
plt.show()
```

```python
# Stacked bargraph
```

```python
plt.bar(df["Students"], df["Math"], label = "Math")
plt.bar(df["Students"], df["Science"], bottom = df["Math"], label = "Math")
bottom_stack = df["Math"] + df["Science"]
plt.bar(df["Students"],df["English"], bottom = bottom_stack, label ="English")
plt.title("Stacked Graph")
plt.xlabel("Student")
plt.ylabel("Total Marks")
plt.legend()
plt.show()
```

```python
#Scatter plot
```

```python
plt.scatter(df["Math"], df["Science"], color = "red", marker = "x")
plt.title("Scatter")
plt.xlabel("Math")
plt.ylabel("Science")
plt.show()
```

```python
#Pie Chart
```

```python
avg = [df["Math"].mean(), df["Science"].mean(), df["English"].mean()]
sub = ["Math", "Science", "English"]
plt.pie(avg, labels = sub, autopct = "%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

```python
# Heat Map
```

```python
a = df[["Math", "Science","English"]].corr()
sna.heatmap(a, annot = True, cmap = "coolwarm")
plt.title("HeatMap")
plt.show()
```
